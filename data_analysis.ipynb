```python
import json
import matplotlib.pyplot as plt
import numpy as np
```

```python
with open('data/dataset_2.json') as f:
    data = json.load(f)
```

```python
l = len(data.get('X'))
print(f"O dataset tem {l} linhas")
```

```python
X = np.array(data['X'])
y = np.array(data['y'])
print(X.shape, y.shape)
```

```python
x0_points = [x for x in X[:, 0]] # Pega todas as linhas e so a coluna zero
x1_points = [x for x in X[:, 1]] # Pega  todas as linhas e a coluna um
```

```python
plt.figure(figsize=(6, 6))
plt.scatter(x0_points, x1_points, c=y, cmap='coolwarm', edgecolor='k', s=20)
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Dataset - distribuição das classes')
plt.colorbar(label='classe')
plt.show()
```
