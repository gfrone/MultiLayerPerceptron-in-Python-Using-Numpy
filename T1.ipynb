```python
import numpy as np
```

```python
class Dense():
    def __init__(self, units : int, initialization : str = "he", activation : str = None):
        self.units = units
        self.activation = activation
        self.initialization = initialization
        
        self.W = None # Pesos  - Inicializar Randomico ou HeInitializaton
        self.b = None # Biases - Inicializar com zeros
        
    def build(self, n_in, rng):
        self.n_in = n_in
        
        if self.activation == "he":
            sigma = np.sqrt(2.0 / n_in)
            self.W = rng.normal(
                0.0,
                sigma,
                size=(n_in, self.units)
            )
            self.b = np.zeros((1, self.units))
            
        elif self.activation == "random":
            std = 0.01 # inicializar com baixa variabilidade
            W = self.rng.normal(0.0, std, size=(self.n_in, self.units))
            b = np.zeros((1, self.units))
            return W, b
        
        else:
            raise ValueError("Escolha uma inicialização válida: "he" ou "random". ")
```
